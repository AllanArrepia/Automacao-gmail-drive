# Automação de Sistemas e Processos com Python

### Vou iniciar a automação importando tudo que vou precisar e já entrando no Google Drive para baixar a planilha de vendas da semana

```python
import pyautogui
import time
import pandas as pd
import pyperclip

pyautogui.press("win")
pyautogui.write("chrome")
pyautogui.press("enter")
time.sleep(2)
pyautogui.click(x=1372, y=660)
time.sleep(1)
pyautogui.click(x=131, y=95)
time.sleep(2)
pyautogui.click(x=626, y=330, button="right")
time.sleep(2)
pyautogui.click(x=749, y=401)
time.sleep(4)
```

### Agora vou ler o arquivo com o pandas e armazenar o faturamento e a quantidade de produtos vendidos na semana

```python
caminho = r"C:\Users\Usuario\Downloads\vendas-teste.xlsx"
tabela = pd.read_excel(caminho)
faturamento = tabela["Valor Total"].sum()
qtde_produtos = tabela["Quantidade"].sum()
```

### Agora vou enviar um email fictício com as informações que salvei

```python
pyautogui.hotkey("ctrl", "t")
pyautogui.click(x=238, y=99)
time.sleep(2)
pyautogui.click(x=81, y=180)
time.sleep(2)
pyautogui.write("pythonimpressionador@gmail.com")
time.sleep(1)
pyautogui.press("enter")
pyautogui.press("tab")
pyperclip.copy("Relatório de Vendas")
pyautogui.hotkey("ctrl","v")
time.sleep(1)
pyautogui.press("enter")
pyautogui.press("tab")
texto = ("""
Prezados,

Segue o relatório de vendas dessa semana.

Faturamento: R${:,.2f}
Quantidade de produtos vendidos: {:,}

Qualquer dúvida estou à disposição.
Abs,
Allan
""".format(faturamento,qtde_produtos))

pyperclip.copy(texto)
pyautogui.hotkey("ctrl","v")
time.sleep(1)
pyautogui.hotkey("ctrl","enter")
```
